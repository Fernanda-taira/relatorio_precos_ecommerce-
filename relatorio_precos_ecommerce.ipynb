```python
import requests
import pandas as pd

print("--- 🚀 INICIANDO PIPELINE DE ETL ---")

# 1. EXTRACT (Extração dos dados brutos)
url_api = "https://fakestoreapi.com/products"
resposta = requests.get(url_api)
dados = resposta.json()
print(f"✅ [Extract] {len(dados)} produtos capturados via API.")

# 2. TRANSFORM (Transformação e limpeza)
produtos_limpos = []
for item in dados:
    produtos_limpos.append({
        'Nome do Produto': item['title'],
        'Categoria': item['category'],
        'Preço (USD)': item['price']
    })

df_produtos = pd.DataFrame(produtos_limpos)
print("✅ [Transform] Dados convertidos com sucesso para o formato de tabela.")

# 3. LOAD (Carga blindada para o Excel em português)
nome_arquivo = 'relatorio_precos_ecommerce.csv'

# A MÁGICA ACONTECE AQUI:
# sep=';' -> Diz ao Excel para separar as colunas por ponto e vírgula
# encoding='utf-8-sig' -> Garante que o 'ç' e os acentos fiquem perfeitos
df_produtos.to_csv(nome_arquivo, index=False, sep=';', encoding='utf-8-sig')

print(f"✅ [Load] Arquivo final salvo com sucesso em: {nome_arquivo}")
print("--- 🎉 PIPELINE FINALIZADO COM SUCESSO ---")
```
